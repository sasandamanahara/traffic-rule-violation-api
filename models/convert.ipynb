```python
import torch

checkpoint = torch.load(
    r"C:\Users\GF63 10SCSR\Desktop\app\traffic-rule-violation-api\models\1.pt",
    map_location="cpu",
    weights_only=False   # allow full unpickling
)
model = checkpoint['model'].float()  # access the model
print(model)

# Example: change first conv layer to accept 3 channels
from torch import nn
old_conv = model.model[0]  # first layer
new_conv = nn.Conv2d(
    in_channels=3,  # change to 3 channels
    out_channels=old_conv.out_channels,
    kernel_size=old_conv.kernel_size,
    stride=old_conv.stride,
    padding=old_conv.padding,
    bias=(old_conv.bias is not None)
)
with torch.no_grad():
    new_conv.weight[:, :3, :, :] = old_conv.weight[:, :3, :, :]  # copy weights

model.model[0] = new_conv
torch.save({'model': model}, "1_rgb.pt")
```
